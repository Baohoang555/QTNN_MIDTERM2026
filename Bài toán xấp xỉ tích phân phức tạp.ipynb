```python
import numpy as np
import matplotlib.pyplot as plt

# số lượng mẫu
n = 100000

# sinh ngẫu nhiên từ U(0,1)
x = np.random.uniform(0, 1, n)

# tính giá trị hàm
f = np.exp(-x**2) * np.cos(x)

# Monte Carlo estimate
est = np.mean(f)

# standard error
se = np.std(f) / np.sqrt(n)

print("Estimated Integral:", est)
print("Standard Error:", se)
```

```python
import numpy as np

# số lượng mẫu
n = 100000

# sinh ngẫu nhiên từ Exp(1)
x = np.random.exponential(scale=1, size=n)

# tính giá trị hàm
f = np.sin(x) / (x + 1)

# Monte Carlo estimate
est = np.mean(f)

# standard error
se = np.std(f) / np.sqrt(n)

print("Estimated Integral:", est)
print("Standard Error:", se)
```

```python
import numpy as np

# số lượng mẫu
n = 100000

# sinh ngẫu nhiên từ N(0,1)
x = np.random.normal(0, 1, n)

# tính giá trị hàm
f = np.log(x**2 + 2) / (1 + x**2)

# Monte Carlo estimate
est = np.sqrt(2*np.pi) * np.mean(f)

# standard error
se = np.sqrt(2*np.pi) * np.std(f) / np.sqrt(n)

print("Estimated Integral:", est)
print("Standard Error:", se)
```
